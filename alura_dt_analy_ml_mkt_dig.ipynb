## Variáveis quantitativas

Nossas variáveis quantitativas serão:

Escolhemos quatro variáveis quantitativas que vamos somar todas as visitas desses usuários. Iremos pegar usuário a usuário e somar. São elas a *newVisits*, que mostra se é a primeira visita do usuário ou não. A segunda é a *pageviews*, que mostra quantas páginas o usuário viu durante a sessão a partir do momento em que ele acessou o site.
A terceira métrica é o *bounces*, que podemos chamar de rejeição. É uma visita que não gera nenhuma interação. Conhecemos como visita que envia apenas uma request. E a *hits*, que é o volume de interação que o Google recebeu do site. Nos ajuda a entender quanto o usuário interagiu no site, mas vai variar de site a site.

## Gradient Boosting

O gradient boosting ao invés de fazer apenas um preditor que vai funcionar para a base inteira, cria vários preditores pequenos e sequenciais. Ele cria um preditor, faz as predições e calcula o erro. O que ele aprende sobre o erro serve de aprendizado para o próximo. É sequencial. E assim por diante. Ele vai criando um modelo que vai funcionar para nossa base.

### Importando o modelo e criando uma instância dele

Na biblioteca Scikit-Learn existe um algoritmo chamado gradiente boosting. É um algoritmo do tipo ensemble. Existem dois tipos, o classifier e o regressor. Nosso modelo é do tipo regressão, porque estamos tentando prever quanto o usuário vai gastar no site. O classifier serve para classificar valores. https://cursos.alura.com.br/course/machine-learning-marketing-digital/task/52126

A ideia do Ensemble Learning é simples: juntar forças para criar algo mais robusto e confiável do que qualquer modelo isolado. E para isso o Ensemble Learning treina vários modelos (que podem ser iguais ou diferentes) para realizar a mesma tarefa. Depois, combina os resultados deles para formar uma “superprevisão”.
https://hub.asimov.academy/blog/o-que-e-ensemble-learning/#:~:text=A%20ideia%20%C3%A9%20simples%3A%20juntar,para%20formar%20uma%20%E2%80%9Csuperprevis%C3%A3o%E2%80%9D.

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gb = GradientBoostingRegressor(random_state=42)
```

### Ajustando (fitting) o modelo nas nossas bases

No aprendizado supervisionado, os algoritmos de Machine Learning ajustam os parâmetros das funções para que elas satisfaçam as condições dadas pelos valores das etiquetas. Esse ajuste é conhecido como "model fit" e é realizado durante o treinamento do modelo com base nos dados. https://pt.stackoverflow.com/questions/377643/o-que-%C3%A9-overfitting-e-underfitting-em-machine-learning#:~:text=No%20aprendizado%20supervisionado%2C%20os%20algoritmos,modelo%20com%20base%20nos%20dados.

```python
gb.fit(X_train,y_train)
```

### Fazendo as predições
